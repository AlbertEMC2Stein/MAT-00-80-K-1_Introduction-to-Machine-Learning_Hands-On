## Get started

1. Open the notebook for the current week from the schedule below.
2. Read the instructions and run code cells with **Shift + Enter**.
3. Work from top to bottom: later cells may depend on earlier ones.
4. Change examples, make predictions, and complete the **Your turn** exercises.
5. Write your observations in the Markdown cells. Explaining a result is part of the analysis.

If you get stuck, ask your instructor or a neighbour, or use the learning-tutor prompt linked below. Errors are a normal part of learning to program.

## Working towards your data project

The sessions build the skills needed to analyse a dataset independently. In your project, document and justify the relevant steps:

- **Data organization:** understand the data and handle issues such as missing values, outliers, and conversions.
- **Preprocessing:** choose appropriate transformations, scaling, and features.
- **Analysis:** use one or two suitable methods, reasonably tune hyperparameters, and use cross-validation when applicable.
- **Presentation:** explain your results and predictions with clear text and useful visualizations.
- **Reflection:** discuss what worked, what did not, alternative methods, model behaviour, and limitations.

Keep asking: *What is our question? What is our baseline? How do we evaluate the result? What can we conclude?*
