## How to study with the notebooks

For the new format, starting with Week 4:

1. **Prepare before class.** Open the week's notebook and work through its explanations and examples. Allow enough time to pause and revisit difficult sections; there is no 90-minute limit for notebook study.
2. **Run complete examples.** The main walkthrough provides full, executable code, with explanations of both the syntax and the analytical choices. You do not need to fill in missing code.
3. **Work from top to bottom.** Run code cells with **Shift + Enter**. Later cells may depend on earlier ones. Compare the outputs with the explanations.
4. **Pause at the self-checks.** Predict a result, interpret an output, or consider an implication before continuing. Record your reasoning in your own words.
5. **Explore when ready.** The final **Your turn** section offers optional opportunities to modify examples and investigate further. Predict what a change will do, then run it and compare.
6. **Bring questions to class.** Note any unclear concept, code expression, or unexpected result. Classroom time builds on your preparation through discussion, clarification, and application.

Weeks 1–3 currently retain the earlier format with code-completion exercises. They are planned for revision using Week 4 as the teaching-style reference.

If you get stuck, use the learning-tutor prompt below for a focused explanation or debugging help. You can ask for help understanding provided code without first writing your own solution. Errors and questions are a normal part of learning.

## Working towards your data project

The sessions build the skills needed to analyse a dataset independently. In your project, document and justify the relevant steps:

- **Data organization:** understand the data and handle issues such as missing values, outliers, and conversions.
- **Preprocessing:** choose appropriate transformations, scaling, and features.
- **Analysis:** use one or two suitable methods, reasonably tune hyperparameters, and use cross-validation when applicable.
- **Presentation:** explain your results and predictions with clear text and useful visualizations.
- **Reflection:** discuss what worked, what did not, alternative methods, model behaviour, and limitations.

Keep asking: *What is our question? What is our baseline? How do we evaluate the result? What can we conclude?*
