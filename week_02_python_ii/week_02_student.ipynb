## Worked example 1 — Repeating operations with loops (about 12 minutes)

Printing every day separately creates repetitive code. It also becomes impractical when a dataset contains many observations.

````python
print(f"{days[0]}: €{sales[0]:.2f}")
print(f"{days[1]}: €{sales[1]:.2f}")
print(f"{days[2]}: €{sales[2]:.2f}")
````

A `for` loop repeats its indented body once for each element. One run of the body is called an **iteration**.

````python
for day in days:
    print(day)
````

`zip()` pairs corresponding values from two lists. This only makes sense here because `days` and `sales` have matching lengths and orders.

````python
for day, sales_value in zip(days, sales):
    print(f"{day}: €{sales_value:.2f}")
````

A loop can contain the conditions from Week 1. Python executes the entire indented block during every iteration.

````python
for day, sales_value in zip(days, sales):
    if sales_value >= daily_target:
        status = "target reached"
    else:
        status = "below target"

    print(f"{day}: €{sales_value:.2f} — {status}")
````

## Worked example 2 — Defining reusable functions (about 11 minutes)

The classification is useful in several places. A **function** gives a name to a reusable operation.

````python
def classify_sales(sales_value, target):
    if sales_value >= target:
        return "target reached"
    return "below target"
````

In the definition, `sales_value` and `target` are **parameters**. In the call below, `510` and `450` are **arguments**. `return` sends the result back to the calling code.

````python
status = classify_sales(510, 450)
print(status)
````

### `return` is not the same as `print`

A returned value can be stored and reused. Printing only displays a value.

````python
def calculate_difference(sales_value, target):
    return sales_value - target

difference = calculate_difference(510, daily_target)
print(f"Difference: €{difference:.2f}")
````

````python
for day, sales_value in zip(days, sales):
    status = classify_sales(sales_value, daily_target)
    print(f"{day}: €{sales_value:.2f} — {status}")
````

## Worked example 3 — Raising and catching errors (about 8 minutes)

A negative sales value is technically a number, but it is not valid for our simplified business example. A function can deliberately **raise** an error instead of returning a misleading result.

````python
def validate_sales(sales_value):
    if sales_value < 0:
        raise ValueError("Sales cannot be negative.")

    return sales_value

print(validate_sales(510))
````

Uncomment the next line, run the cell, and read the final line of the error message. Comment it again before continuing.

Use `try` to attempt an operation and `except ValueError` to handle this specific expected problem. The error is explained rather than silently hidden.

````python
try:
    checked_sales = validate_sales(-50)
    print(f"Checked sales: €{checked_sales:.2f}")
except ValueError as error:
    print(f"Could not use the value: {error}")
````

## Worked example 4 — Creating a `Company` class (about 10 minutes)

So far, the company name, target, days, sales, and functions are separate. A **class** is a blueprint that bundles related data and behaviour. You do not need to write this class from memory. Read it from top to bottom and look for code you already know.

````python
class Company:
    def __init__(self, name, daily_target, days, sales):
        if daily_target <= 0:
            raise ValueError("The daily target must be positive.")

        if len(days) != len(sales):
            raise ValueError("Each day must have exactly one sales value.")

        for sales_value in sales:
            validate_sales(sales_value)

        self.name = name
        self.daily_target = daily_target
        self.days = days
        self.sales = sales

    def classify_sales(self, sales_value):
        if sales_value >= self.daily_target:
            return "target reached"
        return "below target"

    def average_sales(self):
        return sum(self.sales) / len(self.sales)

    def print_sales_report(self):
        print(f"{self.name}\n")

        for day, sales_value in zip(self.days, self.sales):
            status = self.classify_sales(sales_value)
            print(f"{day}:\t€{sales_value:.2f}\t{status}")
````

Create one **object** from the class. From here on, `company` is the central object in our workflow.

````python
company = Company(
    name=company_name,
    daily_target=daily_target,
    days=days,
    sales=sales
)
````

| Concept | Example |
|---|---|
| class | `Company` |
| object or instance | `company` |
| attributes | `company.name`, `company.sales` |
| methods | `company.average_sales()`, `company.print_sales_report()` |

`self` refers to the current object inside the class definition.

````python
print(company.name)
print(f"Target: €{company.daily_target:.2f}")
print(f"Average sales: €{company.average_sales():.2f}")
print()
company.print_sales_report()
````

The class checks its inputs. This prevents the creation of an object with inconsistent business data.

````python
try:
    invalid_company = Company(
        name="Broken Business",
        daily_target=450,
        days=["Monday", "Tuesday"],
        sales=[420, -50]
    )
except ValueError as error:
    print(f"Could not create the company: {error}")
````

## Worked example 5 — NumPy arrays and random data (about 11 minutes)

### A reproducible random number generator

Computer-generated random values are **pseudo-random**. A seed defines a reproducible starting point. The same seed and the same sequence of commands produce the same values.

````python
rng = np.random.default_rng(seed=42)
first_sample = rng.normal(loc=500, scale=70, size=5)

rng = np.random.default_rng(seed=42)
repeated_sample = rng.normal(loc=500, scale=70, size=5)

print(first_sample)
print(repeated_sample)
print(np.array_equal(first_sample, repeated_sample))
````

### Three distributions, three possible meanings

- **Normal:** simulated daily sales around a centre, with variation on both sides.
- **Uniform:** discount rates where every value in an interval can occur.
- **Exponential:** non-negative waiting times, with many short waits and some long ones.

These are simplified assumptions, not facts about the real company.

````python
rng = np.random.default_rng(seed=42)

simulated_sales = rng.normal(loc=500, scale=70, size=30)
discount_rates = rng.uniform(low=0.05, high=0.15, size=30)
waiting_times = rng.exponential(scale=4, size=30)

print("First five simulated sales:", simulated_sales[:5])
print("First five discount rates:", discount_rates[:5])
print("First five waiting times:", waiting_times[:5])
````

## Worked example 6 — Organising the simulation with pandas (about 6 minutes)

A pandas **DataFrame** organises observations in rows and named columns. The usual abbreviation for pandas is `pd`.

````python
import pandas as pd

simulation_data = pd.DataFrame({
    "day": np.arange(1, 31),
    "sales": simulated_sales,
    "discount_rate": discount_rates,
    "waiting_time": waiting_times
})

simulation_data.head()
````

`simulation_data` is an object and `.head()` is a method—just like `company` is an object and `.average_sales()` is a method. Add a Boolean column using the target stored in the central company object.

````python
simulation_data["target_reached"] = (
    simulation_data["sales"] >= company.daily_target
)

print(f"Mean simulated sales: €{simulation_data['sales'].mean():.2f}")
print(f"Days reaching the target: {simulation_data['target_reached'].sum()}")
simulation_data.head()
````
