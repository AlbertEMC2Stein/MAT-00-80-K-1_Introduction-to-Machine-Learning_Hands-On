<h2 id="section-3-creating-readable-text">3. Creating readable text</h2>

Business results need labels and units. A string can be enclosed in double quotation marks (`"..."`) or single quotation marks (`'...'`). The opening and closing marks of one string must match.

Using two styles is helpful when text itself contains quotation marks. A single-quoted string can contain double quotation marks directly, and a double-quoted string can contain an apostrophe directly. If the inner and outer marks are the same, place a backslash before the inner mark to **escape** it.


````python
report_title = "Weekly sales report"
currency = 'euro'

# Different outer quotation marks allow the inner marks to appear directly.
manager_message = 'The manager said, "Check Friday first."'
ownership_message = "Campus Corner's weekly report"

# A backslash escapes the same kind of quotation mark used outside.
escaped_double_quote = "The target is called \"daily sales\"."
escaped_single_quote = 'Campus Corner\'s report'

print(report_title)
print(currency)
print(manager_message)
print(ownership_message)
print(escaped_double_quote)
print(escaped_single_quote)
````

### Multiline strings

Triple quotation marks create a string that spans several source-code lines. Either triple double quotes (`"""..."""`) or triple single quotes (`'''...'''`) can be used. The line breaks become part of the stored string.

Multiline strings are useful for a short report template or a longer block of explanatory text. The opening and closing triple quotation marks still need to match.


````python
weekly_report = """Campus Corner's weekly report
The manager said, "Review Friday first."
Daily target: €450"""

print(weekly_report)
````

In these specifications, the number before the decimal point is the **minimum field width**. It includes the sign, digits, and decimal point. A value is never cut off merely to fit the width. The precision after the decimal point controls the displayed decimal places. `f`, `d`, and `e` select fixed-point, integer, and scientific notation.


### Dynamic width and precision

If the width is stored in a variable, Python can make it dynamic:

- an f-string can nest the width and precision variables: `f"{value:{field_width}.{decimal_places}f}"`;
- classic `%` formatting uses `*` to take the width and precision from the supplied tuple: `"%*.*f" % (field_width, decimal_places, value)`.

In both cases, the values of `field_width` and `decimal_places` determine the display when the code runs.


````python
field_width = 8
decimal_places = 2
format_value = 3.141592653

# Dynamic width and precision with an f-string.
print(f"|{format_value:{field_width}.{decimal_places}f}|")

# The same dynamic choices with classic % formatting.
print("|%*.*f|" % (field_width, decimal_places, format_value))
````

````python
format_value = 3.5
integer_value = 12
large_value = 12345.6

# Vertical bars make spaces and field widths easier to see.
print("|%.2f|" % format_value)
print("|%6.2f|" % format_value)
print("|%06.2f|" % format_value)
print("|%+06.2f|" % format_value)
print("|%d|" % integer_value)
print("|%1.3e|" % large_value)
````

The f-string versions put the same formatting instructions after the colon. F-strings are especially convenient when one line contains labels, several values, or expressions.


````python
print(f"|{format_value:.2f}|")
print(f"|{format_value:6.2f}|")
print(f"|{format_value:06.2f}|")
print(f"|{format_value:+06.2f}|")
print(f"|{integer_value:d}|")
print(f"|{large_value:1.3e}|")
````

Two further formats are particularly useful in business reports:

- `:,.2f` adds a thousands separator and shows two decimal places.
- `:.1%` converts a proportion such as `0.10` to a percentage display such as `10.0%`.

The percentage format multiplies the displayed value by 100 and adds `%`; the stored value remains `0.10`.


````python
annual_sales = 1234567.8
discount_rate = 0.10

print(f"Annual sales: €{annual_sales:,.2f}")
print(f"Discount rate: {discount_rate:.1%}")
print("Stored discount rate:", discount_rate)
````

The outputs are `|+03.50|` and `|-03.50|`. The `+` requests an explicit sign, `0` requests zero-padding, `6` is the minimum total width, and `.2f` requests fixed-point notation with two decimal places.

````python
number_of_days = len(sales)
weekly_sales = sum(sales)
average_daily_sales = weekly_sales / number_of_days

print(f"Recorded days: {number_of_days}")
print(f"Weekly sales: €{weekly_sales:.2f}")
print(f"Average daily sales: €{average_daily_sales:.2f}")
print(f"Lowest daily sales: €{min(sales):.2f}")
print(f"Highest daily sales: €{max(sales):.2f}")
````

An `if` statement executes its indented block when its condition is `True`. An `else` block provides the alternative. The colon begins a block, and four spaces make its contents visible to Python.


````python
selected_day = days[0]
selected_sales = sales[0]

if selected_sales >= daily_target:
    decision = "target reached"
else:
    decision = "below target"

print(f"{selected_day}: €{selected_sales:.2f} — {decision}")
````

An `elif` branch adds another possible outcome. Python checks conditions from top to bottom and executes the first matching branch. Therefore, the most restrictive threshold belongs first.


````python
sales_value = 510

if sales_value >= 550:
    category = "high-sales day"
elif sales_value >= 450:
    category = "target reached"
else:
    category = "below target"

print(f"€{sales_value:.2f}: {category}")
````

<h2 id="section-6-repeating-operations-with-for-loops">6. Repeating operations with <code>for</code> loops</h2>

Writing one statement per day repeats the same structure. This becomes impractical as the number of observations grows.


````python
print(f"{days[0]}: €{sales[0]:.2f}")
print(f"{days[1]}: €{sales[1]:.2f}")
print(f"{days[2]}: €{sales[2]:.2f}")
````

A `for` loop repeats its indented body once for each element. One run of the body is an **iteration**. During each iteration, `day` refers to the next value from `days`.


````python
for day in days:
    print(day)
````

`zip()` pairs values at matching positions. On the first iteration, `day` is `Monday` and `sales_value` is `420`. This works because the two lists have the same length and order.


````python
for day, sales_value in zip(days, sales):
    print(f"{day}: €{sales_value:.2f}")
````

A loop can contain a decision. Python executes the entire indented block for every pair. The second level of indentation belongs to the `if` or `else` block.


````python
for day, sales_value in zip(days, sales):
    if sales_value >= daily_target:
        status = "target reached"
    else:
        status = "below target"

    print(f"{day}: €{sales_value:.2f} — {status}")
````

### Counting within a loop

An **accumulator** stores a result that is updated during repeated operations. Start the counter at zero, increase it only when the condition is true, and print the final value after the loop.


````python
days_reaching_target = 0

for sales_value in sales:
    if sales_value >= daily_target:
        days_reaching_target = days_reaching_target + 1

print(f"Days reaching the target: {days_reaching_target}")
````

**Self-check:** Why is the final `print()` aligned with `for` rather than indented inside it? What would the output mean if it were inside the loop?

Aligned code runs after all iterations and reports the final count. An indented print would show a running total after every day. Both can be useful, but they answer different questions.

A **function** gives a name to a reusable operation. We have already called built-in functions such as `print()`, `len()`, and `sum()`. We can define our own function with `def`.

### `return` and `print()` have different purposes

`print()` displays information. `return` makes a result available to the calling code, so it can be stored, compared, formatted, or used in another calculation.


````python
def print_difference(sales_value, target):
    print(sales_value - target)


def calculate_difference(sales_value, target):
    return sales_value - target


printed_result = print_difference(510, 450)
returned_result = calculate_difference(510, 450)

print("Value returned by print_difference:", printed_result)
print("Value returned by calculate_difference:", returned_result)
````

The first function displays `60` but returns no explicit value, so Python returns `None`. The second function returns `60`, which can be stored and reused.

A function can also accept an entire list. The next function uses only its parameter `values`; it does not depend on the global variable named `sales`.


````python
def calculate_average(values):
    total = sum(values)
    number_of_values = len(values)
    return total / number_of_values


weekly_average = calculate_average(sales)
print(f"Weekly average: €{weekly_average:.2f}")
````

### Raising a meaningful error

A negative sales value is technically a number, but it is invalid in this simplified example. A function can deliberately `raise` a specific error rather than return a misleading result.


````python
def validate_sales(sales_value):
    if sales_value < 0:
        raise ValueError("Sales cannot be negative.")

    return sales_value


checked_sales = validate_sales(510)
print("Checked sales:", checked_sales)
````

Use `try` for the operation that may fail and `except ValueError` for the expected problem. The variable `error` contains the message created by `raise`.


````python
try:
    checked_sales = validate_sales(-50)
    print(f"Checked sales: €{checked_sales:.2f}")
except ValueError as error:
    print(f"Could not use the value: {error}")
````

**Self-check:** Why catch `ValueError` instead of every possible error?

Catching the expected type documents what the code knows how to handle. An unrelated error, such as a misspelled variable name, should remain visible so it can be corrected rather than being mistaken for invalid sales data.


<h2 id="section-9-bundling-data-and-behaviour-in-a-class">9. Bundling data and behaviour in a class</h2>

The company name, target, days, sales, and functions currently exist separately. A **class** is a blueprint that describes related data and behaviour together. An **object** is one concrete value created from that blueprint.

The `Company` class below checks its inputs, stores them as attributes, and provides methods for repeated business calculations.


````python
class Company:
    def __init__(self, name, daily_target, days, sales):
        if daily_target <= 0:
            raise ValueError("The daily target must be positive.")

        if len(days) != len(sales):
            raise ValueError("Each day must have exactly one sales value.")

        for sales_value in sales:
            validate_sales(sales_value)

        # Attributes store data belonging to the object.
        self.name = name
        self.daily_target = daily_target
        self.days = days.copy()
        self.sales = sales.copy()

        # Variables with a leading underscore are considered "private" by convention.
        # They are not enforced as private, but the underscore signals to other programmers 
        # that they should not be accessed directly.
        self._longest_day_name_length = max(len(day) for day in days)

        # Using two underscores at the start of a variable name invokes name mangling, which makes 
        # it harder to access the variable from outside the class. It still can be accessed, but
        # it is not recommended. Moreover, some tools may flag an access to such variables as a warning or error
        # to discourage their use outside the class.
        self.__internal_variable = "This is an internal variable."

    def classify_sales(self, sales_value):
        if sales_value >= self.daily_target:
            return "target reached"

        return "below target"

    def total_sales(self):
        return sum(self.sales)

    def average_sales(self):
        return self.total_sales() / len(self.sales)

    def days_reaching_target(self):
        count = 0

        for sales_value in self.sales:
            if sales_value >= self.daily_target:
                count = count + 1

        return count

    def print_sales_report(self):
        print(f"{self.name}\n")

        for day, sales_value in zip(self.days, self.sales):
            status = self.classify_sales(sales_value)
            # The `>` in the format specifier right-aligns the day name.
            print(f"{day:>{self._longest_day_name_length}}: €{sales_value:.2f}\t{status}")
````

<h2 id="section-10-interpret-the-complete-result">10. Interpret the complete result</h2>

Our question was: *How can Python help us describe, check, and report a small shop's weekly sales?*

The final cell calculates a small set of results from the company object. Each line has one purpose: quantify the week, compare it with the target, and identify the strongest recorded day. The list method `.index(value)` returns the position of the first matching value, allowing us to retrieve the day at the same position.


````python
total = company.total_sales()
average = company.average_sales()
difference_from_target = average - company.daily_target
best_sales = max(company.sales)
# Find the position of the highest value, then use the same position for the day.
best_index = company.sales.index(best_sales)
best_day = company.days[best_index]

print(f"Total sales: €{total:.2f}")
print(f"Average daily sales: €{average:.2f}")
print(f"Average minus target: €{difference_from_target:.2f}")
print(f"Best recorded day: {best_day} (€{best_sales:.2f})")
````

### Reading the result

- Total sales for the five recorded days were **€2,420**.
- Average daily sales were **€484**, which is **€34 above** the €450 target.

These calculations describe the five values supplied to Python. They do not explain why Friday was strongest, predict a future week, or establish profitability. We would need more observations and information about products, customers, promotions, and costs for those questions.
