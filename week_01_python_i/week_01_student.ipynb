## Worked example 2 — Data types and comparisons (about 10 minutes)

Values have different **types**. Today we use four important types:

- `int`: a whole number, such as `18`
- `float`: a number with a decimal point, such as `3.50`
- `str`: text inside quotation marks, such as `"Monday"`
- `bool`: either `True` or `False`

Use `type()` to inspect a value.

```python
day = "Monday"
customers = 37
average_purchase = 11.35
shop_is_open = True

print(type(day))
print(type(customers))
print(type(average_purchase))
print(type(shop_is_open))
```

Comparison operators ask questions and return a Boolean value. Notice the difference between assignment (`=`) and comparison (`==`).

| Question | Operator |
|---|---|
| equal to | `==` |
| not equal to | `!=` |
| greater / smaller than | `>` / `<` |
| at least / at most | `>=` / `<=` |

```python
monday_sales = 420
daily_target = 450

print(monday_sales > daily_target)
print(monday_sales == 420)
print(day != "Friday")
```

## Worked example 3 — A week of sales (about 15 minutes)

A **list** stores several values in a fixed order. Square brackets create a list; commas separate its elements.

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
sales = [420, 380, 510, 460, 620]

print(days)
print(sales)
```

### Indexing

Each element has a position called an **index**. Python starts counting at zero.

| Element | Monday | Tuesday | Wednesday | Thursday | Friday |
|---|---:|---:|---:|---:|---:|
| Index | 0 | 1 | 2 | 3 | 4 |

The index `-1` selects the last element.

```python
print("First day:", days[0])
print("Wednesday's sales:", sales[2])
print("Last sales value:", sales[-1])
```

Lists can be changed. Suppose Tuesday's sales were entered incorrectly: the correct value is €395.

```python
sales[1] = 395
sales
```

Python has useful built-in functions for lists of numbers. Here, `len()` counts values, `sum()` adds them, and `min()` and `max()` find the smallest and largest values.

```python
number_of_days = len(sales)
weekly_sales = sum(sales)
average_daily_sales = weekly_sales / number_of_days

print("Number of days:", number_of_days)
print("Weekly sales:", weekly_sales)
print("Average daily sales:", average_daily_sales)
print("Lowest daily sales:", min(sales))
print("Highest daily sales:", max(sales))
```

## Worked example 4 — Simple decisions (about 12 minutes)

An `if` statement runs code only when a condition is `True`. Code inside the statement is indented by four spaces.

```python
friday_sales = 620
daily_target = 450

if friday_sales >= daily_target:
    print("The daily target was reached.")
else:
    print("The daily target was not reached.")
```

Change `friday_sales` to `400` and run the cell again. Which branch is executed now?

An optional `elif` branch allows more than two outcomes. Python checks the conditions from top to bottom and executes the first matching branch.

```python
sales_value = 510

if sales_value >= 550:
    print("High-sales day")
elif sales_value >= 450:
    print("Target reached")
else:
    print("Below target")
```
