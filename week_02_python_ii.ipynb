<h2 id="section-1-packages">1. What are packages?</h2>

Python provides a core language: values, variables, conditions, loops, functions, and classes all belong to that language. It also provides reusable collections of code for specialised tasks. Instead of writing numerical array operations, table tools, or plotting systems ourselves, we can use code developed and tested for those purposes.

A **package** is an organised collection of reusable Python code. Packages can provide functions, classes, constants, and smaller components called **modules**. In everyday discussion, the words package and library are sometimes used broadly; the important practical idea is that they add capabilities to the Python environment.

### Installing and importing are different steps

A package must first be **installed** in the Python environment. Installation places its code where Python can find it. The course environment installs the packages listed in `requirements.txt` when Binder is built, so no installation command is needed during the notebook.

An `import` statement then makes an installed package available in the current kernel session. Restarting the kernel clears imported names from memory, so the import cells must be run again. Importing does not reinstall the package.

Python also includes a **standard library**. Its modules are distributed with Python and do not require a separate package installation. For example, the standard-library `math` module provides mathematical functions.

````python
import math

# The module name before the dot identifies where sqrt() comes from.
square_root = math.sqrt(81)
print("Square root of 81:", square_root)
````

### Names and aliases

The dot in `math.sqrt()` separates the module name from a name stored inside it. This is a **namespace**: it makes the source of a function explicit and prevents unrelated packages from placing every name in the same space.

An alias shortens a frequently used package name. `import numpy as np` means that this notebook will refer to NumPy through `np`. The aliases `np`, `pd`, and `plt` are widely used conventions, but they are still ordinary names created by the import statements.

**Self-check:** What is the difference between “NumPy is installed” and “NumPy is imported”?

Installed means that NumPy's code is available in the Python environment. Imported means that the current running session has loaded it and created a name through which our code can access it.

<h2 id="section-2-numpy-numerical-arrays">2. NumPy: numerical arrays</h2>

**NumPy** is a package for numerical computing. Its central object is the **N-dimensional array**, written `ndarray`. An array stores values in a regular shape and supports calculations on many values at once.

Week 1 used Python lists. Lists are flexible: one list can contain different kinds of values. NumPy arrays are designed for regular numerical data and normally store one data type. This regular structure makes numerical operations concise and efficient.

### Importing NumPy

A package must be imported before its objects and functions are available. `import numpy as np` assigns the conventional short alias `np`. The alias does not change NumPy; it only gives us a shorter name to write.

````python
import numpy as np

print("NumPy version:", np.__version__)
````

### Creating an array from a list

`np.array(...)` converts a compatible sequence into an array. The following five values are the Campus Corner sales already used in Week 1. We use them briefly to connect the familiar list with the new array object.

````python
sales_list = [420, 395, 510, 475, 620]
sales_array = np.array(sales_list)

print("Python list:", sales_list)
print("NumPy array:", sales_array)
print("List type:", type(sales_list))
print("Array type:", type(sales_array))
````

Every array has attributes that describe its structure. Attributes belong to an object and are accessed with a dot, as in `sales_array.shape`. Unlike a method call, an attribute is not followed by parentheses.

The comma in `(5,)` marks a one-element tuple. Here it means one dimension of length five. The exact integer type can differ between systems.

````python
print("Dimensions:", sales_array.ndim)
print("Shape:", sales_array.shape)
print("Number of values:", sales_array.size)
print("Stored data type:", sales_array.dtype)
````

**Self-check:** Which expression is a method call: `sales_array.shape` or `sales_array.mean()`?

`shape` is an attribute that stores structural information. `mean()` is a method that performs a calculation, so it uses parentheses. This object–attribute–method notation continues the class concepts from Week 1.

### Calculating with a complete array

An operation between an array and one number is applied **element by element**. This style is called **vectorised calculation**. It replaces many explicit loops when the same numerical operation applies to every value.

Suppose we want to investigate a scenario in which every recorded sales value is 5% higher. Multiplying the array by `1.05` returns a new array; it does not change `sales_array`.

````python
higher_sales = sales_array * 1.05
difference = higher_sales - sales_array

print("Original sales:", sales_array)
print("5% higher:", higher_sales)
print("Increase per day:", difference)
````

The list-based equivalent needs a loop and an accumulator. Both versions are valid; the array expression states the numerical transformation more directly and is significantly faster for large datasets.

````python
higher_sales_list = []

for value in sales_list:
    higher_sales_list.append(value * 1.05)

print(higher_sales_list)
````

### Indexing and slicing

Array indexing starts at zero, just like list indexing. A **slice** selects a range. In `start:stop`, the start is included and the stop is excluded. Leaving out the start means “from the beginning”; leaving out the stop means “to the end.”

````python
print("First value:", sales_array[0])
print("Last value:", sales_array[-1])
print("First three values:", sales_array[:3])
print("Values at indices 1, 2, and 3:", sales_array[1:4])
````

A NumPy slice usually creates a **view** of the same underlying data rather than an independent copy. Changing the view can therefore change the original array. Use `.copy()` when an independent array is required.

The next cell demonstrates the difference and then restores the original value.

````python
first_two_view = sales_array[:2]
first_two_copy = sales_array[:2].copy()

first_two_view[0] = 430
first_two_copy[1] = 999

print("Original after changing the view:", sales_array)
print("Independent copy:", first_two_copy)

# Restore the teaching data for later cells.
sales_array[0] = 420
````

### Boolean masks

Comparing an array with one value produces one Boolean result per element. This Boolean array is called a **mask**. Placing the mask inside square brackets selects only positions where the mask is `True`.

````python
daily_target = 450
target_mask = sales_array >= daily_target
sales_reaching_target = sales_array[target_mask]

print("Sales values:", sales_array)
print("Target mask:", target_mask)
print("Values reaching the target:", sales_reaching_target)
print("Number reaching the target:", target_mask.sum())
````

**Self-check:** Why does `target_mask.sum()` count target-reaching days?

In numerical calculations, `True` behaves like 1 and `False` like 0. Summing the mask therefore counts its `True` values. The mask still represents one decision for every original observation.

### Summary methods

Methods such as `.sum()`, `.mean()`, `.min()`, `.max()`, and `.std()` reduce many values to a summary. The **standard deviation** measures how spread out values are around their mean. A value near zero indicates very similar observations; a larger value indicates more variation.

````python
print(f"Total: €{sales_array.sum():.2f}")
print(f"Mean: €{sales_array.mean():.2f}")
print(f"Minimum: €{sales_array.min():.2f}")
print(f"Maximum: €{sales_array.max():.2f}")
print(f"Standard deviation: €{sales_array.std():.2f}")
````

### Two-dimensional arrays and axes

A two-dimensional array can represent a rectangular table. Here, rows represent weeks and columns represent weekdays. The shape `(2, 5)` means two rows and five columns.

An **axis** tells NumPy which dimension to reduce:

**Self-check:** Why does `two_weeks.mean(axis=0)` return five values, while `mean(axis=1)` returns two?

Reducing axis 0 combines the two rows but leaves the five columns. Reducing axis 1 combines the five columns but leaves the two rows. Always connect an axis to the meaning of the array dimensions, not only to its number.

### Creating numerical sequences

`np.arange(start, stop, step)` creates regularly spaced values and excludes the stop value. `np.linspace(start, stop, number)` creates a requested number of evenly spaced values and includes both endpoints by default.

````python
day_numbers = np.arange(1, 8)
discount_scenarios = np.linspace(0.00, 0.20, 5)

print("Day numbers:", day_numbers)
print("Discount scenarios:", discount_scenarios)
````

### Reproducible random values

Simulations use random-looking values to investigate possible outcomes. NumPy's random values are **pseudo-random**: an algorithm generates a sequence from an initial state. `np.random.default_rng(seed=...)` creates a random number generator.

A fixed **seed** makes a particular sequence reproducible. This allows another person to run the same analysis and obtain the same teaching example. Consecutive calls to one generator still produce different values because the generator advances through its sequence.

````python
rng = np.random.default_rng(seed=42)
first_draw = rng.normal(loc=500, scale=70, size=5)
second_draw = rng.normal(loc=500, scale=70, size=5)

repeated_rng = np.random.default_rng(seed=42)
repeated_first_draw = repeated_rng.normal(loc=500, scale=70, size=5)

print("First draw:", first_draw.round(2))
print("Second draw:", second_draw.round(2))
print("Repeated first draw:", repeated_first_draw.round(2))
print("First draw reproduced:", np.array_equal(first_draw, repeated_first_draw))
````

A **probability distribution** describes which values a random process can generate and how frequently different regions occur. Parameters control the distribution; they should be chosen from subject knowledge or estimated from suitable data, not selected only because they create an attractive result.

<table align="center">
<thead><tr><th>Generator method</th><th>Parameters used here</th><th>Possible teaching interpretation</th><th>Useful limitation</th></tr></thead>
<tbody>
<tr><td><code>uniform()</code></td><td>lower and upper bound</td><td>discount rate within an interval</td><td>treats equal-width parts of the interval as equally likely</td></tr>
<tr><td><code>normal()</code></td><td>mean and standard deviation</td><td>daily sales varying around a centre</td><td>unbounded; can generate impossible negative values</td></tr>
<tr><td><code>exponential()</code></td><td>mean scale</td><td>waiting time between customer arrivals</td><td>assumes many short waits and a long right tail</td></tr>
</tbody>
</table>

````python
rng = np.random.default_rng(seed=42)
sample_size = 1000

uniform_discounts = rng.uniform(low=0.05, high=0.15, size=sample_size)
normal_sales = rng.normal(loc=500, scale=70, size=sample_size)
exponential_waits = rng.exponential(scale=4, size=sample_size)

print("First five uniform discounts:", uniform_discounts[:5].round(3))
print("First five normal sales values:", normal_sales[:5].round(2))
print("First five exponential waits:", exponential_waits[:5].round(2))
````

````python
print(f"Uniform discounts — min: {uniform_discounts.min():.3f}, max: {uniform_discounts.max():.3f}")
print(f"Normal sales — mean: {normal_sales.mean():.2f}, std: {normal_sales.std():.2f}")
print(f"Exponential waits — mean: {exponential_waits.mean():.2f}, max: {exponential_waits.max():.2f}")
````

The sample summaries are close to the requested settings, but they are not required to equal them exactly. Random samples vary. A larger sample often gives summaries closer to the generating parameters, but it does not make an unrealistic assumption realistic.

**Self-check:** If `normal(loc=20, scale=30, ...)` represents the number of customers, what problem may occur?

A normal distribution is unbounded and could generate negative customer counts. We would also need integer counts. The choice of distribution and any correction such as clipping must be explained because it changes the simulated process.

<h2 id="section-3-pandas-labelled-tables">3. pandas: labelled tables</h2>

**pandas** is a package for labelled tabular data. A **DataFrame** is a two-dimensional table with rows and named columns. A **Series** is one labelled one-dimensional sequence, often one DataFrame column.

NumPy arrays emphasise regular numerical structure. DataFrames can combine numerical, textual, Boolean, and date columns while preserving labels that state what each value means.

### Creating a DataFrame

The conventional alias for pandas is `pd`. Passing a dictionary to `pd.DataFrame(...)` creates one column from each key. All columns must contain the same number of rows.

````python
import pandas as pd

transactions = pd.DataFrame({
    "product": ["Notebook", "Pen", "Folder", "Notebook", "Pen", "Calculator"],
    "category": ["Paper", "Writing", "Paper", "Paper", "Writing", "Electronics"],
    "units": [18, 30, 12, 15, 25, 4],
    "unit_price": [3.50, 1.80, 2.40, 3.50, 1.80, 18.00],
})

transactions
````

### Inspect before calculating

Before analysis, inspect the table's size, labels, types, and example rows. `head()` returns the first five rows by default. `info()` prints column types and non-missing counts.

````python
print("Shape:", transactions.shape)
print("Columns:", transactions.columns.tolist())
print("\nData types:\n", transactions.dtypes)
print("\nInformation:")
transactions.info()

transactions.head()
````

### Selecting columns

One column label inside brackets returns a Series. A list of labels inside brackets returns a DataFrame. The double brackets in the second expression mean: select the list `['product', 'units']` from the DataFrame.

````python
units_series = transactions["units"]
product_and_units = transactions[["product", "units"]]

print("One-column type:", type(units_series))
print("Two-column type:", type(product_and_units))

product_and_units
````

### Selecting rows and columns with `loc` and `iloc`

- `.loc[row_labels, column_labels]` selects by labels.
- `.iloc[row_positions, column_positions]` selects by integer positions.

The colon means “all” along that dimension. Unlike normal Python slicing, label slices in `.loc` include their final label. Position slices in `.iloc` exclude the stop position.

````python
print("Row label 2, selected columns:")
print(transactions.loc[2, ["product", "units"]])

print("\nFirst three rows and first two columns by position:")
print(transactions.iloc[:3, :2])
````

### Filtering rows

A comparison on a Series creates a Boolean Series. Using it inside `.loc[...]` keeps rows where the value is `True`. Parentheses make each condition visible. Use `&` for element-wise “and” and `|` for element-wise “or.”

````python
many_units = transactions["units"] >= 15
selected_rows = transactions.loc[many_units]

paper_with_many_units = transactions.loc[
    (transactions["category"] == "Paper") & (transactions["units"] >= 15)
]

print("Rows with at least 15 units:")
print(selected_rows)
print("\nPaper rows with at least 15 units:")
paper_with_many_units
````

### Creating derived columns

A derived column combines information already present in the table. pandas applies arithmetic between aligned columns row by row, so no explicit loop is needed.

Here, revenue is units multiplied by unit price. The calculation describes recorded revenue before costs or returns; it is not profit.

````python
transactions["revenue"] = transactions["units"] * transactions["unit_price"]
transactions["large_line"] = transactions["revenue"] >= 50

transactions
````

### Sorting values

`sort_values()` returns rows ordered by one or more columns. `ascending=False` places the largest value first. Unless the result is assigned or `inplace=True` is used, the original DataFrame order remains unchanged.

````python
transactions_by_revenue = transactions.sort_values("revenue", ascending=False)
transactions_by_revenue[["product", "units", "revenue"]]
````

### Summary statistics and counts

`describe()` calculates common summaries for numerical columns. `value_counts()` counts how often each distinct value occurs. These methods describe the available table; they do not explain why a pattern exists.

````python
print("Numerical summary:")
print(transactions[["units", "unit_price", "revenue"]].describe().round(2))

print("\nCategory counts:")
print(transactions["category"].value_counts())
````

### Missing values

Real tables often contain unknown values. pandas represents many missing numerical values with `NaN`, meaning “not a number.” Missing does not automatically mean zero. Replacing an unknown price with zero would state that the product was free, which is a different claim.

We create a copy and deliberately remove two values so that the original table remains complete.

````python
transactions_missing = transactions.copy()
transactions_missing.loc[2, "unit_price"] = np.nan
transactions_missing.loc[4, "category"] = None

transactions_missing
````

````python
print("Missing values by column:")
print(transactions_missing.isna().sum())

print("\nRows containing at least one missing value:")
transactions_missing.loc[transactions_missing.isna().any(axis=1)]
````

There is no universally correct treatment. `dropna()` removes affected rows; `fillna(...)` replaces missing entries. Both choices can change results and should be justified from the meaning of the data.

For illustration, we fill the missing price with the median recorded price and the missing category with `Unknown`. The **median** is the middle ordered value and is less affected by an unusually large price than the mean. This is an illustrative choice, not evidence that these were the true missing values.

````python
transactions_filled = transactions_missing.copy()
median_price = transactions_filled["unit_price"].median()

transactions_filled["unit_price"] = transactions_filled["unit_price"].fillna(median_price)
transactions_filled["category"] = transactions_filled["category"].fillna("Unknown")
transactions_filled["revenue"] = (
    transactions_filled["units"] * transactions_filled["unit_price"]
)

transactions_filled
````

### Grouped summaries

`groupby()` divides rows according to a category, applies a calculation within each group, and combines the results. This is often described as **split–apply–combine**.

The next expression selects the numerical columns to summarise and then calculates a sum for every category. By using `(` and `)` we can break the expression across multiple lines for readability.

````python
# 1. Select "category" column for grouping and only retain the "units" and "revenue" columns
# 2. Calculate the sum of "units" and "revenue" for each category
# 3. Sort the resulting DataFrame by "revenue" in descending order
category_totals = (
    transactions
    .groupby("category")[["units", "revenue"]] 
    .sum()
    .sort_values("revenue", ascending=False)
)

category_totals
````

**Self-check:** Does the category with the most units necessarily have the largest revenue?

No. Revenue depends on both quantity and price. Four calculators can produce more revenue than many low-priced pens. A grouped total answers a descriptive question about this table; it does not show which category is more profitable because costs are absent.

### Storing a DataFrame

A DataFrame exists in the running Python session. To keep a table for later use or share it, write it to a file. The method `to_csv(path, index=False)` stores the rows and column names as comma-separated text. `index=False` omits the DataFrame's row labels: the default index here only numbers the rows and is not a business variable. If the index contains meaningful identifiers, preserve it or turn it into a column before saving.

The example writes a copy of `transactions` to a temporary folder and reads it back. `TemporaryDirectory()` creates a separate folder for this demonstration and removes it when the `with` block ends. To keep a file, provide a suitable persistent path instead and download it before closing Binder: Binder sessions are temporary. Writing to an existing path replaces the existing file, so choose the destination carefully.

````python
from tempfile import TemporaryDirectory

# Create a temporary destination for this demonstration.
with TemporaryDirectory() as folder:
    csv_path = Path(folder) / "transactions.csv"

    # Save column names and data, without the default row index.
    transactions.to_csv(csv_path, index=False)

    # Read the file back while the temporary folder exists.
    restored_transactions = pd.read_csv(csv_path)
    print("Saved and reloaded shape:", restored_transactions.shape)
    print(restored_transactions.head())
````

CSV stores text, so it does not preserve every pandas data type or formatting choice. Inspect the reloaded table, especially dates, identifiers, and missing values. Other formats such as Excel and Parquet have different capabilities and requirements.

**Self-check:** What would happen if the default index were saved and then read as an ordinary CSV column?

The row numbers would appear as an additional column unless the reader explicitly identifies that column as the index. Whether to keep them depends on whether those labels carry useful information.

<h2 id="section-4-matplotlib-communicating-with-charts">4. Matplotlib: communicating with charts</h2>

**Matplotlib** creates charts. A chart should answer a specific question and make the relevant comparison visible. Decoration cannot compensate for an unclear question, unsuitable data, or missing units.

We use Matplotlib's explicit object-oriented style. `plt.subplots()` creates two objects:

The plural word `Axes` names one plotting object; it is different from the x-axis and y-axis drawn inside it.

````python
import matplotlib.pyplot as plt
````

### Line chart: values across an ordered sequence

A line chart is suitable when the horizontal order is meaningful, such as days or months. Connecting points suggests continuity or progression, so a line should not be used merely because it looks familiar.

The method calls configure one `Axes`: `plot()` draws data, `set_xlabel()` and `set_ylabel()` add units and meanings, `set_title()` states the focus, and `legend()` explains labelled elements.

````python
week_numbers = np.arange(1, 7)
weekly_revenue = np.array([2350, 2480, 2410, 2590, 2630, 2710])

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(week_numbers, weekly_revenue, marker="o", label="Revenue")
ax.set_xlabel("Week")
ax.set_ylabel("Revenue in euros")
ax.set_title("Weekly revenue")
ax.grid(alpha=0.3)
ax.legend()
plt.show()
````

### Bar chart: comparing categories

A bar chart compares quantities across discrete categories. A zero baseline is usually important because viewers compare bar lengths. The category order should support the question—for example, descending revenue makes the largest category easy to identify.

````python
category_names = category_totals.index.to_numpy()
category_revenue = category_totals["revenue"].to_numpy()

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(category_names, category_revenue, color="tab:blue")
ax.set_xlabel("Product category")
ax.set_ylabel("Revenue in euros")
ax.set_title("Revenue by product category")
ax.set_ylim(bottom=0)
plt.show()
````

### Histogram: the distribution of one numerical variable

A histogram divides a numerical range into intervals called **bins** and counts observations in each bin. It helps reveal centre, spread, skewness, gaps, and unusual values. It does not preserve the observation order.

The chosen number of bins affects the visible shape. Too few bins hide detail; too many can make random noise look important.

````python
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(normal_sales, bins=20, edgecolor="white", color="tab:green")
ax.axvline(normal_sales.mean(), color="black", linestyle="--", label="Sample mean")
ax.set_xlabel("Simulated daily sales in euros")
ax.set_ylabel("Number of observations")
ax.set_title("Distribution of simulated daily sales")
ax.legend()
plt.show()
````

A scatter plot places one point per observation. Its horizontal and vertical positions represent two numerical variables. A visible association does not by itself establish that one variable causes the other: both may be affected by additional factors.

<h2 id="section-5-complete-campus-corner-workflow">5. Complete Campus Corner workflow</h2>

We now combine all three packages around the shop from Week 1.

**Question:** *Under a simple set of assumptions, what could six weeks of Campus Corner activity look like, and how often might daily sales reach the €450 target?*

This is a **simulation**, not a prediction fitted to historical data. Its conclusions describe the generated example under our assumptions. The purpose is to practise a transparent workflow:

```text
question and assumptions
→ NumPy generation and calculation
→ pandas organisation and summaries
→ Matplotlib visualisation
→ cautious interpretation
```

### Step 1 — State the business settings and assumptions

The company name and target match Week 1. We simulate 42 consecutive days. The assumed process is deliberately simple.

<table align="center">
<thead><tr><th>Quantity</th><th>Assumption</th><th>Reason for including it</th></tr></thead>
<tbody>
<tr><td>customers</td><td>normal variation around 42, plus 8 on weekends</td><td>creates changing daily demand</td></tr>
<tr><td>average purchase</td><td>normal variation around €12, clipped at €4</td><td>keeps purchases positive in this example</td></tr>
<tr><td>discount rate</td><td>uniformly between 0% and 12%</td><td>represents simple daily promotions</td></tr>
<tr><td>waiting time</td><td>exponential with mean 4 minutes</td><td>illustrates a positive, right-skewed duration</td></tr>
</tbody>
</table>

These assumptions are not estimated from Campus Corner records. Changing them changes the result.

````python
company_name = "Campus Corner"
daily_target = 450
number_of_days = 42
simulation_seed = 2026

dates = pd.date_range(start="2026-01-05", periods=number_of_days, freq="D")
day_type = np.where(dates.dayofweek >= 5, "Weekend", "Weekday") # np.where(...) is a vectorized conditional function. If the condition is True, the first value is used; if False, the second value is used.

print("First date:", dates[0].date())
print("Final date:", dates[-1].date())
print("Number of simulated days:", len(dates))
````

### Step 2 — Generate and calculate with NumPy

We create one generator and use it for the complete simulation. `np.where(...)` supplies a weekend increase where the condition is true and zero otherwise. `np.rint(...)` rounds customer counts to the nearest whole number; `.astype(int)` stores them as integers.

`np.clip(values, minimum, maximum)` restricts values to a range. Passing `None` as the maximum leaves the upper side unrestricted. Clipping is a modelling decision: values that would fall below the minimum are placed exactly at the boundary.

````python
rng = np.random.default_rng(seed=simulation_seed)

weekend_customer_increase = np.where(day_type == "Weekend", 8, 0)
customers = rng.normal(loc=42, scale=7, size=number_of_days) + weekend_customer_increase
customers = np.rint(customers).astype(int)
customers = np.clip(customers, 5, None)

average_purchase = rng.normal(loc=12.0, scale=1.8, size=number_of_days)
average_purchase = np.clip(average_purchase, 4.0, None)

discount_rate = rng.uniform(low=0.00, high=0.12, size=number_of_days)
waiting_time_min = rng.exponential(scale=4, size=number_of_days)

# Gross sales are reduced by the simulated discount rate.
sales = customers * average_purchase * (1 - discount_rate)
sales = np.round(sales, 2)

print("First five customer counts:", customers[:5])
print("First five sales values:", sales[:5])
print("All customer counts positive:", (customers > 0).all())
print("All sales values positive:", (sales > 0).all())
````

### Step 3 — Organise the observations with pandas

Every array has length 42, so each position can become one DataFrame row. A row now represents one simulated day. Labels preserve the meaning and unit of each column.

````python
shop_data = pd.DataFrame({
    "date": dates,
    "day_name": dates.day_name(),
    "day_type": day_type,
    "customers": customers,
    "average_purchase_eur": average_purchase,
    "discount_rate": discount_rate,
    "waiting_time_min": waiting_time_min,
    "sales_eur": sales,
})

shop_data.head(n=7) # Specify n=7 to show the first seven rows instead of the default five.
````

Create columns that answer the business question and make reporting easier. `target_reached` compares every row with the same target. `sales_per_customer` is derived from two existing columns. The displayed discount percentage is more readable, while the original proportion remains available for calculations.

````python
shop_data["target_reached"] = shop_data["sales_eur"] >= daily_target
shop_data["sales_per_customer_eur"] = (
    shop_data["sales_eur"] / shop_data["customers"]
)
shop_data["discount_percent"] = 100 * shop_data["discount_rate"]

shop_data.head().round(2)
````

### Step 4 — Check and summarise before plotting

A chart should not be the first data-quality check. Confirm the shape, missing values, and plausible ranges. The assertions below document conditions that must be true for this simulated table. If one is false, Python stops with an `AssertionError` rather than allowing the workflow to continue silently.

````python
print("Shape:", shop_data.shape)
print("Missing values:\n", shop_data.isna().sum())

assert len(shop_data) == number_of_days
assert shop_data["customers"].gt(0).all()
assert shop_data["discount_rate"].between(0.00, 0.12).all()
assert shop_data["sales_eur"].gt(0).all()

shop_data[["customers", "average_purchase_eur", "discount_percent",
           "waiting_time_min", "sales_eur"]].describe().round(2)
````

A grouped summary compares weekdays and weekends. We calculate means rather than totals because the two groups contain different numbers of days. The difference reflects our simulation design, which explicitly added weekend customers; it is not discovered evidence about the real shop.

````python
day_type_summary = (
    shop_data
    .groupby("day_type")[["customers", "sales_eur", "waiting_time_min"]]
    .mean()
    .round(2)
)

day_type_summary
````

### Step 5 — Visualise time, distribution, and relationships

No single chart answers every question. We create three panels:

1. a line chart for sales across dates and the target;
2. a histogram for the distribution of daily sales; and
3. a scatter plot for customers and sales, separating weekdays from weekends.

The plots use the same DataFrame but emphasise different structures.

````python
fig, axes = plt.subplots(1, 3, figsize=(18, 5), layout="constrained")

# Panel 1: sales over time with the business target.
axes[0].plot(shop_data["date"], shop_data["sales_eur"], marker="o", markersize=3)
axes[0].axhline(daily_target, color="red", linestyle="--", label="Daily target")
axes[0].set_title("Simulated sales over time")
axes[0].set_xlabel("Date")
axes[0].set_ylabel("Sales in euros")
axes[0].tick_params(axis="x", rotation=45)
axes[0].legend()

# Panel 2: distribution of the 42 daily sales values.
axes[1].hist(shop_data["sales_eur"], bins=10, edgecolor="white", color="tab:green")
axes[1].axvline(shop_data["sales_eur"].mean(), color="black", linestyle="--", label="Mean")
axes[1].set_title("Distribution of daily sales")
axes[1].set_xlabel("Sales in euros")
axes[1].set_ylabel("Number of days")
axes[1].legend()

# Panel 3: one point per day, with day types shown separately.
for label, color in [("Weekday", "tab:blue"), ("Weekend", "tab:orange")]:
    subset = shop_data.loc[shop_data["day_type"] == label]
    axes[2].scatter(subset["customers"], subset["sales_eur"],
                    label=label, color=color, alpha=0.8)

axes[2].set_title("Customers and daily sales")
axes[2].set_xlabel("Customers")
axes[2].set_ylabel("Sales in euros")
axes[2].legend()

fig.suptitle(f"{company_name}: simulated six-week overview", fontsize=14)
plt.show()
````

### Step 6 — Calculate the result used in the interpretation

We calculate the target count and a few descriptive quantities directly from the table. This keeps the written interpretation tied to the actual executed values.

````python
target_days = int(shop_data["target_reached"].sum())
target_share = shop_data["target_reached"].mean()
mean_sales = shop_data["sales_eur"].mean()
median_sales = shop_data["sales_eur"].median()
best_row = shop_data.loc[shop_data["sales_eur"].idxmax()]

print(f"Target reached: {target_days} of {number_of_days} days ({target_share:.1%})")
print(f"Mean daily sales: €{mean_sales:.2f}")
print(f"Median daily sales: €{median_sales:.2f}")
print(f"Largest simulated sales value: €{best_row['sales_eur']:.2f} on {best_row['date'].date()}")
````

### Interpreting the simulated result

The output answers a narrow question: under this seed and these stated assumptions, how often do the generated sales values meet €450, and what does their distribution look like?

The line chart identifies variation over the 42 simulated dates and shows the target as a common reference. The histogram shows how the sales values are distributed but removes their order. The scatter plot shows that higher customer counts tend to accompany higher sales in this simulation. That relationship is partly built into the formula `customers × average purchase × (1 − discount)`.

We cannot conclude that Campus Corner will achieve this result, that weekends are truly stronger, or that changing discounts will cause a particular sales outcome. We assumed a weekend customer increase and generated all values rather than estimating a process from real records. A different seed changes individual days; different distribution parameters change the process itself.

**Self-check:** Which change is more fundamental: using a new seed or increasing the assumed weekend customer boost from 8 to 15?

A new seed produces a different sample from the same assumed process. Changing the weekend boost changes the assumed process. Both alter the output, but the second changes the model of the business rather than only its random realisation.
